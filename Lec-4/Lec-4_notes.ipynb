## Supervised Learning:   
Given a data set of input-ouput pairs, learn a function to map inputs to outputs.   

### **Classification:** Task of learning a function mapping an input point to a discrete category
   - Weather: Rainy or Cloudy. Based on historical data, look for pattern. We can structure the data: Date, Humidity, Pressure, Label (Rain or no Rain).
   - Given some $f(humidity, pressure)$, we now results $\{Rain, No Rain\}$. We need to figure out $f$.
     - We attempt to estimate a func. $h$, and goal is to make $h$ as close as $f$.
   - Example: Red: Not Rainy, Blue: Rainy   
      
      <img src="image.png" alt="classification" width="30%" height="auto">   
   - **Nearest-neighbor Classification:** choose category based on nearest neighbor.
   - **$k$-Nearest-neighbor Classification:** choose category based on $k$-nearest neighbors.
   - **Linear Regression:** Linearly separate the categories, as long as it does a pretty good job.
     - $x_{1}$: Humidity
     - $x_{2}$: Pressure
     - Hypothesis Function: 
      
      $$
      h( x_1 , x_2 ) =
      \begin{cases}
         \text{Rain, $1$.} & \text{if } w_{0} + w_{1}x_{1} + w_{2}x_{2} \geq 0\\
         \text{No Rain, $0$.} & \text{Otherwise.}
      \end{cases}
      $$
     - Here $w_{1}, w_{2}$ puts importance in Humidity and Pressure, and $w_{0}$ is just another weight to get different function.
     - _Weight vector_ $W = \{ w_{0}, w_{1}, w_{2}\}$. Goal is to learn this.
     - _Input vector_ $x = \{ 1, x_{1}, x_{2} \}$. Represents a datapoint.
     - $\vec{W} \cdot \vec{x} =  w_{0} + w_{1}x_{1} + w_{2}x_{2}$.
     - Then general hypothesis is:
      $$
      h_{w}(X) =
      \begin{cases}
         1 & \text{if } \mathbf{W} \cdot \mathbf{X} \geq 0\\
         0 & \text{Otherwise.}
      \end{cases}
      $$


**Perception Learning Rule:**   
- Given data point $(x,y)$, update each weight according to: $w_{i} = w_{i} + \alpha(y-h_{w}(\mathbf{X})) \times x_{i}$.
- $w_{i} = w_{i} + \alpha(\text{actual value}-\text{estimate}) \times x_{i}$
- If $actual = estimate$, then weight $w_{i}$ doesn't change. If $actual > estimate$, then we are increasing the $w_{i}$, if $actual < estimate$, we are decreasing the $w_{i}$
- _Note, by doing this we are effectively creating a Threshold function._
