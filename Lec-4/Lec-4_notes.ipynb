## Supervised Learning:   
Given a data set of input-ouput pairs, learn a function to map inputs to outputs.   

### **Classification:** Task of learning a function mapping an input point to a discrete category
   - Weather: Rainy or Cloudy. Based on historical data, look for pattern. We can structure the data: Date, Humidity, Pressure, Label (Rain or no Rain).
   - Given some $f(humidity, pressure)$, we now results $\{Rain, No Rain\}$. We need to figure out $f$.
     - We attempt to estimate a func. $h$, and goal is to make $h$ as close as $f$.
   - Example: Red: Not Rainy, Blue: Rainy   
      
      <img src="image.png" alt="classification" width="30%" height="auto">   
   - **Nearest-neighbor Classification:** choose category based on nearest neighbor.
   - **$k$-Nearest-neighbor Classification:** choose category based on $k$-nearest neighbors.
   - **Linear Regression:** Linearly separate the categories, as long as it does a pretty good job.
     - $x_{1}$: Humidity
     - $x_{2}$: Pressure
     - Hypothesis Function: 
      
      $$
      h( x_1 , x_2 ) =
      \begin{cases}
         \text{Rain, $1$.} & \text{if } w_{0} + w_{1}x_{1} + w_{2}x_{2} \geq 0\\
         \text{No Rain, $0$.} & \text{Otherwise.}
      \end{cases}
      $$
     - Here $w_{1}, w_{2}$ puts importance in Humidity and Pressure, and $w_{0}$ is just another weight to get different function.
     - _Weight vector_ $W = \{ w_{0}, w_{1}, w_{2}\}$. Goal is to learn this.
     - _Input vector_ $x = \{ 1, x_{1}, x_{2} \}$. Represents a datapoint.
     - $\vec{W} \cdot \vec{x} =  w_{0} + w_{1}x_{1} + w_{2}x_{2}$.
     - Then general hypothesis is:
      $$
      h_{w}(X) =
      \begin{cases}
         1 & \text{if } \mathbf{W} \cdot \mathbf{X} \geq 0\\
         0 & \text{Otherwise.}
      \end{cases}
      $$


**Perception Learning Rule:**   
- Given data point $(x,y)$, update each weight according to: $w_{i} = w_{i} + \alpha(y-h_{w}(\mathbf{X})) \times x_{i}$.
- $w_{i} = w_{i} + \alpha(\text{actual value}-\text{estimate}) \times x_{i}$
- If $actual = estimate$, then weight $w_{i}$ doesn't change. If $actual > estimate$, then we are increasing the $w_{i}$, if $actual < estimate$, we are decreasing the $w_{i}$
- _Note, by doing this we are effectively creating a Threshold function._

  <img src="image-1.png" alt="Thresholdfunc" width="30%" height="auto">  

- But this kind of hard thresholding doesn't give us anything in between.
- For example, if the classification line keeps a datapoint in "Rainy" blue zone, doesn't mean that we are fully sure if it belongs there. For example point "A" in the image would be classified as "Rainy" but we might not be sure.
- Also you cannot take any derivative of this kind of threshold func.

   <img src="image-2.png" alt="Thresholdfunc_2" width="30%" height="auto">  

- So we do "Soft Thresholding". Like this:

### Support Vector Machine (SVM)
There are many lines we can draw to separate two set of datapoints. But we would like to select a line that generalizes the classification better than others.  
- So SVM is technique that tries to find **Maximum Margin Separator**.
  - Boundary that maximizes the distance between any of the (groups of) data points.
- There are groups of data where no linear line exists that can fully separate the groups. But SVM can work on higher dimension to find a curved line to still separate the groups.  
   <img src="image-4.png" alt="SVM_1" width="30%" height="auto">  

### Regression:   
- Supervised learning task of learning a function mapping an input point to a continuous value (not a discrete value like "Rain" or "Not Rain").
- Eample:
  - I a data that where input is money spent of advertising, and output is total sales. $f(advert.) = sales$.
  - Hypothesis $h(advert.) = sales$. We would like learn $h$ so that $h \approx f$.   
   <img src="image-5.png" alt="linreggress" width="30%" height="auto">   

**Loss Function:**
- Function that expresses how poorly our hypothesis performs.
- _Discrete Category:_ Loss function: **0-1 loss function**. 
- $L_{1}$-_Loss Function:_ goals is to minimize following.
  - $L(actual, predicted) = |actual - predicted|$.   
   
   <img src="image-6.png" alt="l1-loss" width="30%" height="auto">   

- $L_{2}$-_Loss Function:_
  - $L(actual, predicted) = (actual-predicted)^2$.
  - Use it where we would like to minimize error for outlier cases.


**Overfitting:**
- A model that fits too closely to a particular data set and therefor may fail to generalize to future data.   
   <img src="image-7.png" alt="overfitting" width="30%" height="auto">  

- In optimization we minimize $cost(h)$ of hypothesis $h$.
- If $cost(h) = loss(h)$, then in order to minimize $loss(h)$ we try to minimize the $cost(h)$ by same amount. This could lead to _Overfitting_.
- **Regularization:** _prenalizing hypothesis that are more complex to favor simpler, more general hypothesis._
  - So instead we do this: $cost(h) = loss(h) + \lambda complexity(h)$. Here $\lambda$ is a weight to decide how much importance we will give to the $complexity(h)$.
- **Holdout Cross-validation:** _data splitting to train and test sets._
- **$k$-fold Cross-validation:** _Splitting data into $k$ sets and experimenting $k$ times, using each set as a test set once, and using the remaining data as training set._
