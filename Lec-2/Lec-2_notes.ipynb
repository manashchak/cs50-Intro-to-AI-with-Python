# Uncertainty  

## Probability:
- Possible world $\omega$.
- Probability: $0 \leq P(\omega) \leq 1$.
- $\sum_{\omega \in \Omega}P(\omega) = 1$.   

### Unconditional Probability:
Consider, a pair of dice rolls (two dices rolled together.) There are 36 equally likely scenarios:

<img src="image.png" alt="Dice Scenarios" width="50%" height="auto"/>   

But sum of the two dice rolls for each scenario are not equally likely,   

<img src="image-1.png" alt="Dice Sum" width="50%" height="auto"/>   

- $P(\text{sum to 12}) = \frac{1}{36}$.
- $P(\text{sum to 7}) = \frac{6}{36} = \frac{1}{6}$.
