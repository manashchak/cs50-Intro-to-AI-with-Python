# Uncertainty  

## Probability:
- Possible world $\omega$.
- Probability: $0 \leq P(\omega) \leq 1$.
- $\sum_{\omega \in \Omega}P(\omega) = 1$.   

### Unconditional Probability:
Consider, a pair of dice rolls (two dices rolled together.) There are 36 equally likely scenarios:

<img src="image.png" alt="Dice Scenarios" width="50%" height="auto"/>   

But sum of the two dice rolls for each scenario are not equally likely,   

<img src="image-1.png" alt="Dice Sum" width="50%" height="auto"/>   

- $P(\text{sum to 12}) = \frac{1}{36}$.
- $P(\text{sum to 7}) = \frac{6}{36} = \frac{1}{6}$.

### Conditional Probability:
**This is what we use most of the time in the topic of AI or Machine Learning.**   
Degree of belief in a proposition given some evidence that has already been revealed. $P(a|b) = ?$.   
- **Bayes' Rule:**
  - $P(a|b) = \frac{P(a \& b)}{P(b)} = \frac{P(b|a)P(a)}{P(b)}$.


### Random Variable:   
- Roll = {1,2,3,4,5,6}. Here Roll is a random variable, and {1,2,3,4,5,6} are all the possible values.
- **Probability Distribution:**
  - Sums to 1.
  - Vector Form: $P(Flight) = \langle 0.6, 0.3, 0.1 \rangle$.   

### Independence:
- The knowledge that one event occurs does not affect the prob of the other event.
- $P(a \& b) = P(a)P(b|a) = P(b)P(a|b) = P(a)P(b)$.

### Prob. Rules:
- **Negation Rule:** P(&not; $\alpha$) = 1 - P($\alpha$).
- **Inclusion-Exclusion:** P(a &or; b) = P(a) + P(b) - P(a &and; b).
- **Marginalization:** P(a) = P(a,b) + P(a, &not;b). ["," typically means "and"]
  - Adding two disjointed joint prob distributions, we can figure out P(a).
  - $P(X=x_{i}) = \sum_{j} P(X=x_{i}, Y=y_{j})$.   
  <img src="image-3.png" alt="MarginalProb" width="50%" height="auto"/>    

- **Conditioning:**
  - P(a) = P(a|b)P(b) + P(a|&not;b)P(&not;b).
  - $P(X=x_{i}) = \sum_{j} P(X=x_{i}| Y=y_{j}) P(Y=y_{j})$.



## Bayesian Network
- Data structure that represents the dependencies between random variables.
- Directed graph.
- Each node is a random variable.
- $X \rightarrow Y$, X is a parent of Y.
- Each node X has prob distribution. $P(X|parent(X))$.

<img src="image-4.png" alt="BayesianNetwork" width="50%" height="auto"/>   


  - "Appointment" depends on "Train", which depends on "Maintenance" and "Rain", and so on.
  - P(light rain, no maintenance, delayed train) = P(light rain).P(no maintenance|light rain).P(delayed train|no maintenance, light rain).

# Inference   
- Query X: variable for which to compute distribution.
- Evidence variable E: observed variables for event e. 
- Hidden variables Y: non-ovidencce, non-query variable.
- Goal: Calculate $P(X|e)$.   

#### Example:
- P(Appointment | light, no) = ?
  - P(Appointment | light, no) = $\alpha$ P(Appointment, light, no).  Here $\alpha$ is proportional constant.
  - $\alpha$ P(Appointment, light, no) = $\alpha$ [ P(Appointment, light, no, on time) + P(Appointment, light, no, delayed) ].

- **Inference by Enumeration:**
  - _Conditional probability is proportional to the joint probability._
  - _Then we can marginalize the joint probability over all the hidden variables $y$._
  - $P(X|e) = \alpha P(X,e) = \alpha \sum_{y} P(X, e, y)$.
