## Simulated Annealing:   
- Let's consider we started at a particular state and we want to find global maxima.
- If we follow Hill Climbing Algo, we might go right and get stuck on a local maxima.
- So it is sometime a good idea to allow our optimization algo to go in the bad direction. This is the idea of Simulated Annealing, that multiple random good and bad decision might end up anealate each other and reach our goal.
- _At the begining we have higher "Temperature" state, meaning we are more likely to accept bad decision. As the optimization moves along, we will decrease the "Temperature", meaning that we will slow down on how many bad decision we will accept_.

<img src="image-2.png" alt="SimAnnealing" width="50%" height="auto"/>  


> Sudo Code: **Simulated Annealning**:   
> ```
> function Simulated-Annealing(problem, max):
>   current = init state of prob.
>   for t=1 to max:
>       T = Temperature(t)
>       neighbor = random neighbor of current
>       $\Delta E$ = how much better neighbor is than current (+ve means better, -ve means worse)
>       if $\Delta E$ > 0:
>           current = neighbor
>       else:
>           with probability $e^{\Delta E / T}$ set current = neighbor
>   return current
> ```

### Linear Programing:   
- Family of Problems that might have following:
  - Minimize a cost function: $c_{1}x_{1} + c_{2}x_{2} + \dots + c_{n}x_{n}$.
  - With constraint of form: $a_{1}x_{1} + a_{2}x_{2} + \dots + a_{n}x_{n} \leq b$, or of form $a_{1}x_{1} + a_{2}x_{2} + \dots + a_{n}x_{n} = b$.
  - With bounds for each variables $l_{i} \leq x_{i} \leq u_{i}$.

**Example:**   
- Two machines $X_{1}$ and $X_{2}$. $X_{1}$ costs 50/hrs to run. $X_{2}$ costs 80/hrs to run. Goal is to minimize cost.
  - cost func: $50x_{1} + 80x_{2}$. Minimize this.
- Constraints:
  - $X_{1}$ requires 5 units of labor per hours. $X_{2}$ requires 2 units of labor per hour. Total of 20 units of labor to spend.
    - $5x_{1} + 2x_{2} \leq 20$.
  - $X_{1}$ produces 10 units of output per hour. $X_{2}$ produces 12 units of output per hour. Company needs 90 units of output.
    - $10x_{1} + 12x_{2} \geq 90$. ===> $-10x_{1} - 12x_{2} \leq -90$.

### Constraint Satisfaction:
- Consider following problem: where students are taking multiple courses.
- Problem: Schedule exams during the exam slots, so that not students should have to take two exams in same day.

<img src="image-5.png" alt="ExamsProb" width="50%" height="auto"/>  

First, we can create a constraint graph to represent, that the courses have constraints, meaning some of these courses cannot happen in the same day.   

<img src="image-6.png" alt="ExamsProb-Graph" width="50%" height="auto"/>

- Set pf variables: $\{ X_{1}, X_{2}, \dots, X_{n} \}$.
- Set of domains for each variables: $\{ D_{1}, D_{2}, \dots, D_{n} \}$.
- Set of constraints $C$.

<img src="image-7.png" alt="ExamsProb-Setup" width="50%" height="auto"/>

- **Hard Constraints:** Typically must met.
- **Soft Constraints:** indicates some sort of preferrences.

- **Unary Constraint:** Single variable. $\{ A \neq Monday \}$.
- **Binary Constraint:** Two variable. $\{ A \neq B \}$.

- **Node Consistency:** When all the values in a variable's domain satisfy the variable's unary constraints.
  - Meaning, between two domain, the variables and the unary constraint should not conflict.

- **Arc Consistency:** When all the values in a variable's domain satisfy the variable's _binary_ constraints.
  - To make $X$ arc-consistent with respect to $Y$, remove elements from $X$'s domain until every choice for $X$ has a possible choice for $Y$.

Through process of elimination to satisfy Node and Arc consistency, here is the final results between $A$ and $B$,

<img src="image-8.png" alt="ExamsProb-Consistency" width="50%" height="auto"/>   

> Sudo Code: Arc Consistency:
> ```
> function REVISE(csp, X, Y): csp= constraint satisfaction problem
>   revised = false
>   for x in X.domain:
>       if no y in Y.domain satisfies constraint for (X,Y):
>           delete x from X.domain
>           revised = true
>   return revised
> ```

More general algo: AC-3, inforces arc consistency to entire problem

> Sudo Code: AC-3(csp)
> ```
> queue = all arcs in csp
> while queue non-empty:
>   (X,Y) = DEQUEUE(queue)
>   if REVISE(csp, X, Y):
>       if size of X.domain == 0:
>           return false
>       else:
>           for each Z in X.neighbors - {Y}:
>               ENQUEUE(queue, (Z,X))
> return true
> ```
